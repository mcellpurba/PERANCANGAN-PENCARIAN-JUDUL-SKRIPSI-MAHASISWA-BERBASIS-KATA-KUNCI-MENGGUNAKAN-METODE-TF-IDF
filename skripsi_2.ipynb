```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

from Sastrawi.Stemmer.StemmerFactory import StemmerFactory
from Sastrawi.StopWordRemover.StopWordRemoverFactory import StopWordRemoverFactory
```

```python
documents = [
    "Sistem Informasi Akademik Berbasis Web",
    "Analisis Sentimen Media Sosial Menggunakan Machine Learning",
    "Perancangan Sistem Temu Kembali Informasi",
    "Sistem Rekomendasi Produk Menggunakan Collaborative Filtering",
    "Implementasi Data Mining untuk Prediksi Mahasiswa",
    "Sistem Informasi Perpustakaan Digital",
    "Penerapan Natural Language Processing untuk Chatbot",
    "Analisis Klasifikasi Menggunakan Decision Tree",
    "Sistem Pencarian Dokumen Menggunakan TF IDF dan Cosine Similarity",
    "Perancangan Aplikasi E Learning Berbasis Web"
]
```

```python
# Stemmer & Stopword
factory = StemmerFactory()
stemmer = factory.create_stemmer()

stop_factory = StopWordRemoverFactory()
stopword = stop_factory.create_stop_word_remover()

def preprocess(text):
    text = text.lower()
    text = stopword.remove(text)
    text = stemmer.stem(text)
    return text

processed_docs = [preprocess(doc) for doc in documents]

print("Hasil Preprocessing:")
for i in range(len(documents)):
    print(f"{documents[i]} → {processed_docs[i]}")
```

```python
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(processed_docs)

print("\nShape TF-IDF:", tfidf_matrix.shape)
```

```python
query = "sistem informasi"   # kamu bisa ganti bebas
query_processed = preprocess(query)

print("\nQuery:", query)
print("Query setelah preprocessing:", query_processed)
```

```python
query_vector = vectorizer.transform([query_processed])
similarity = cosine_similarity(query_vector, tfidf_matrix)

ranking = similarity[0]

sorted_results = sorted(
    enumerate(ranking),
    key=lambda x: x[1],
    reverse=True
)
```

```python
print("Cosine Similarity Scores:")
for i, score in enumerate(ranking):
    print(f"Document {i+1}: {score:.4f}")
```

```python
print("\n===== HASIL PENCARIAN =====")

top_n = 5

for i, (idx, score) in enumerate(sorted_results[:top_n]):
    print(f"\nRanking {i+1}")
    print(f"Skor: {score:.4f}")
    print(f"Dokumen: {documents[idx]}")
```
